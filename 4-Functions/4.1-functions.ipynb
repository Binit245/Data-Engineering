```python
## why functions?

num =24

if num%2==0:
    print("the number is even")
else:
    print("the number is odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")
```

```python
## Call this function
even_or_odd(24)
```

```python
## function with multipple parameters

def add(a,b):
    return a+b
result = add(2,4)
print(result)
```

```python
## Default Parameters

def greet(name):
    print(f"Hello {name} Welcome to paradise")


greet("Krish")
```

```python
def greet(name="Guest"):
    print(f"Hello {name} Welcome to paradise")


greet("Binit")
```

```python
## Variable Length Arguments
## Positional And Keywords arguments


## Positional Arguments

def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Binit")
```

```python
## Keywords Arguments

## In the kryword arguments all the paramaeters will be in the form of key value pairs.
## and it wrotes in (**kwargs) form 

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Binit",age=32,country="India")
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f" Positional argument :{val}")

    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"Binit",name="Binit",age=32,country="India")
```

```python
## Return Statements

def multiply(a,b):
    return a*b

multiply(2,3)
```

```python
## Return Multiple Parameters

def multiply(a,b):
    return a*b,a

multiply(2,3)
```
