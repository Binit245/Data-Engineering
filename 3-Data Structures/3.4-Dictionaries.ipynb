Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in Key-value pairs. Keys must be unique and immutable (e.g., Strings, numbers, or tuples), while values can be of any type.

```python
## Creating dictionaries
empty_dict = {}
print(type(empty_dict) )
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"name":"Binit", "age": 25, "grade": 24}
print(student)
print(type(student))
```

```python
## Single Key is always used  

## this will overwrite the previous value of the key
student = {"name":"Binit", "age": 32, "name": 24}
print(student)
```

```python
## Accessing Dictonary Elements

## Accessing the value of a key in a dictionary is done by using square brackets and the key name.
student = {"name":"Binit", "age": 25, "grade": 'A'}
print(student['grade'])

## Accessing using get() method
print(student.get('age'))
print(student.get('address'))  ## returns None if key is not found
print(student.get('address', 'Not Found'))  ## returns Not Found if key is not found
```

```python
## Modifying Dictonary Elements

## dictionary are mutable, so you can add, update or delete elements.
print(student)
```

```python
student["age"]=33  ## updated the value of key "age"
print(student)
student["address"]="Kathmandu"  ## added new key and value
print(student)
```

```python
## dictionaries Methods

keys = student.keys()  ## returns all the keys in the dictionary
print(keys)
values = student.values()  ## returns all the values in the dictionary
print(values)

items = student.items()  ## returns all the key-value pairs in the dictionary as tuples
print(items)
```

```python
## shallow copy of dictionary
## it creates a new dictionary with the same key-value pairs as the original dictionary, but the two dictionaries are independent of each other.
student_copy = student.copy()
print(student)
print(student_copy)
```

```python
student["name"]="Ram"  ## updated the value of key "name"
print(student)
print(student_copy)  ## student_copy remains unchanged
```

```python
student_copy1 = student.copy()  ## creates a shallow copy of the dictionary
print(student_copy1)
print(student)
```

```python
student["name"] = "mohan" ## updated the value of key "name"
print(student_copy1) ## student_copy1 remains unchanged
print(student) ## student is updated
```

```python
## Iterating over a dictionary
## you can use loop to iterate over Dictionaries, Keys, Values, Items

## Iterating over Keys
for keys in student.keys():
    print(keys)
```

```python
## Iterate over Values
for value in student.values():
    print(value)
```

```python
## Iterate over key value pairs
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
## Nested Dictionaries
## A nested dictionary is a dictionary that contains another dictionary as a value. You can access the values of the inner dictionary using the keys of the outer and inner dictionaries.

students={
    "student1":{"name": "krish", "age": 32},
    "student2":{"name": "peter", "age": 35}
}
print(students)
```

```python
## Access nested Dictionaries elements
print(students["student1"]["age"]) ## Accessing the age of student1
print(students["student2"]["name"]) ## Accessing the name of student2
```

```python
students.items()
```

```python
## Iterating over nested Dictionaries

for student_id, student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key, value in student_info.items():
        print(f"{key}: {value}")
```

```python
## Dictionary Comprehension
## Dictionary comprehension is a concise way to create dictionaries in Python. It allows you to create a new dictionary by applying an expression to each item in an iterable, such as a list or another dictionary.

squares = {x: x**2 for x in range(5)}
print(squares)
```

```python
## conditional dictionary comprehension

even = {x: x**2 for x in range(10) if x%2 == 0}
print(even)
```

```python
## Practical Examples

## Use a Dictionary to count the frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}
for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## Merge 2 Dictionaries into one

dict1={"a":1, "b":2}
dict2={"b":3, "c":4}
merged_dict={**dict1, **dict2}
print(merged_dict)
```
