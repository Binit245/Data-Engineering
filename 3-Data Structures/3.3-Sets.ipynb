# Sets


Sets are a built-in data type in python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performung mathematical set operations like union, intersection, difference, and symmetric difference

```python
## Create a set
my_set = {1, 2, 3, 4, 5}
print(my_set) 
print(type(my_set))  # Output: <class 'set'>
```

```python
my_empty_set = set()
print(type(my_empty_set)) 
```

```python
my_set = set([1,2,3,4,5,6])
print(my_set) 
```

```python
my_set = set([1,2,3,4,5,6,5,4,2])
print(my_set) 
```

```python
## Basic set operaions
## Adding and Removing Elements
my_set.add(7)
print(my_set)  
```

```python
## Removing the elements from a set
my_set.remove(3) 
print(my_set)
```

```python
my_set.remove(10)  # This will raise a KeyError since 10 is not in the set
```

```python
my_set.discard(10)  # Does not raise an error if the element is not found
print(my_set)
```

```python
## pop method
removed_element = my_set.pop()  # Removes and returns an arbitrary element
print(removed_element)
print(my_set)
```

```python
## clear alll the elements from the set
my_set.clear()
print(my_set)  # Output: set()
```

```python
## set Membership test
## membership test is used to check if an element is present in a set or not. It returns True if the element is present, otherwise it returns False.
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)  
print(6 in my_set) 
```

```python
## Mathematical set operations
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

## Union
union_set = set1.union(set2)
print(union_set)  

## Intersection
intersection_set = set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)  ## Updates set1 with the intersection of set1 and set2
print(set1)
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
## Difference
print(set1.difference(set2))  ## Returns a new set with elements in set1 but not in set2
set1.difference_update(set2)  ## Updates set1 with the difference of set1 and set2
```

```python
set2.difference(set1)  
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print(set1.symmetric_difference(set2))  ## Returns a new set with elements in either set1 or set2 but not in both
# set1.symmetric_difference_update(set2)  ## Updates set1 with the symmetric difference of set
```

```python
## sets methods
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5}

## issubset() method
print(set1.issubset(set2))  ## Returns True if set1 is a subset of set2, otherwise False
print(set1.issuperset(set2))  ## Returns True if set1 is a superset of set2, otherwise False
```

```python
lst = [1,2,2,3,4,4,5]
set(lst)  ## Converts the list to a set, removing duplicates
```

```python
## Counting unique words in text

text = "In this tutorial we are discussing about sets"
words = text.split()

## convert list of words to get unique words
unique_words = set(words)
print(unique_words)
print(len(unique_words))  ## Count of unique words
```
