Introduction to Tuples

Explanation:

Tuples are ordered collections of items that are immutable. They are similar to lists, but their immutability makes them different

```python
## Tuple Methods
print(numbers.count(1))  ## This will return the number of times 1 appears in the tuple.
print(numbers.index(3))  ## This will return the index of the first occurrence of 3 in the tuple.
```

```python
## Packing and Unpacking Tuples 
## Packing a tuple means creating a tuple with multiple values, while unpacking a tuple means extracting the values from a tuple and assigning them to variables.


## Packing a tuple
packed_tuple = 1, 2, "Hello", 3.4
print(packed_tuple)
```

```python
## Unpacking a tuple

a,b,c,d = packed_tuple

print(a)
print(b)
print(c)
print(d)
```

```python
## Unpacking with *
## The * operator can be used to unpack a tuple into multiple variables, where one variable can capture multiple values as a list.
numbers = (1, 2, 3, 4, 5, 6)
first,*middle,last = numbers
print(first)  ## This will print the first element of the tuple.
print(middle)  ## This will print the middle elements of the tuple as a list.
print(last)  ## This will print the last element of the tuple.
```

```python
## Nested Tuples

## Nested List
lst = [[1,2,3,4],[5,6,7,8],[1,"Hello",3.14,"c"]]
lst[0][0:3]
```

```python
lst = [[1,2,3,4],[5,6,7,8],(1,"Hello",3.14,"c")]
lst[2][0:3]
```

```python
nested_tuples = ((1,2,3), ("a","b","c"),(True, False))

## access the elements inside the tuple
print(nested_tuples[0])  ## This will print the second element of the first tuple, which is 2.
print(nested_tuples[1][2])
```

```python
## iterating over nested tuples
for sub_tuple in nested_tuples:
    for item in sub_tuple:
        print(item,end=" ")
    print()  ## This will print each element of the nested tuples in a new line.
```

### Conclusion

Tuples are versatile and useful in many real-world scenarios where an immutable and ordered collection of items is required. They are commonly used in data structures, function arguments and return values, and as dictionary keys.
