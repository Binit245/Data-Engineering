# Variables 

Variables are fundamental elements in programming used to store data that can be referenced and manipluated in a program. In python, variables are created when you assign a value to them, and they do not need explicit declaration to reserve memory space. The decleration happens automatically when you assign a value to a variable.

```python
## Declaring And Assigning Variables

age =32
hieght=5.8
name="Binit"
is_student=True

## Printing The Variables
print("age:",age)
print("hieght:",hieght)
print("name:",name)
print("is_student:",is_student)
```

```python
## Naming Conventions
## Variables names should be descriptive and meaningful
## They must always start with a letter or an underscore '_'.
## Variables names case sensitive, which means that 'age' and 'Age' are two different variables.

## Valid Variable names

first_name="Binit"
last_name="Kumar"
```

```python
## Understanig Variables types
## Python is dynamically typed, typed of a variable is determined at runtime and you don't need to declare it explicitly.
age=25  # int
height=5.7 # float
name="Binit" # str
is_student=True # bool

print(type(age))
print(type(height))
print(type(name))  
print(type(is_student)) 
```

```python
## Dynamic Typing
## Python allows the type of variable to change as the program executes. This is known as dynamic typing. For example, you can assign an integer to a variable and later assign a string to the same variable.

var=10 # int
print(var,type(var))

var="Hello" # string
print(var,type(var))

var=3.14 # float
print(var,type(var))
```

```python
## input

age=input("Enter your age: ")
print(age,type(age))  # The input function returns a string, so the type of 'age' will be <class 'str'>.

age=int(input("Enter your age: "))
print(age,type(age))  # Now, 'age' will be of type <class 'int'> because we converted the input to an integer using int().
```

```python
##  Simple Calculator
num1=float(input("Enter first number:"))
num2=float(input("Enter second number:"))

sum = num1+num2
difference = num1-num2
product = num1*num2
quotient = num1/num2 

print("Sum:",int(sum))
print("Difference:",difference)
print("Product:",product)
print("Quotient:",quotient)
```
