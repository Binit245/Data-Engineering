```python
## Arithmetic Operation

a=10
b=5
add_result = a+b # addition
sub_result = a-b # subtraction
mul_result = a*b # multiplication (returns a float if either operand is a float)
div_result = a/b # division (returns a float)
floor_div_result = a//b # floor division (returns the largest integer less than or equal to the division result)
modulus_result = a%b # modulus (remainder)
exponent_result = a**b # exponent 

print(add_result)
print(sub_result)
print(mul_result)
print(div_result)
print(floor_div_result)
print(modulus_result)
print(exponent_result)
```

```python
## Simple Calculator

num1= float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Performing arithmetic operations

addition = num1+num2
substraction = num1-num2
multiplication = num1*num2
division = num1/num2
floor_division = num1//num2
modulus = num1%num2
exponent = num1**num2

 # Displaying the results

print("Addition: ", addition)
print("Subtraction: ", substraction)
print("Multiplication: ", multiplication)
print("Division: ", division)
print("Floor Division: ", floor_division)
print("Modulus: ", modulus)
print("Exponent: ", exponent)
```
