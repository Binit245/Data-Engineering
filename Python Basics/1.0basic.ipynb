Syntax refers to the set of rules that defines the combinations of symbols that are considered to be correctly structured programs in a language. In simpler terms, Syntax is about the correct arrangement of words and symbols in a code.

Semantics refers to the meaning or the interpretation of the symbols, characters, and commands in a language. It is about what the code is supposed to do when it runs.

```python
## Basic Syntax Rules In Python
## Case Sensitivity - Python is case sensitive.

name="John"
Name="naik"

print(name) 
print(Name) 

```

## Indentation
Indenntation in python is used to define the structure and hierarchy of the code. Unlike many other programming languages that use braces {} to delimits block of code, Python uses indentation to detrmine the grouping of statements. This means that all the statements within a block must be indented at the same level.

```python
## Indentation - Python uses indentation to define blocks of code. Consistent use of spaces (comonly 4) or a tab is required.

age =32
if age>30:
    print(age)

print(age)  # This line is not indented, so it is outside the if block and will always execute.
```

```python
## This is a single line comment
print("Hello, World!")  # This is an inline comment
```

```python
## Line Continuation - Python allows you to break long lines of code into multiple lines for better readability. You can use the backslash (\) character to indicate that a line should continue onto the next line.    

total = 1 + 2 + 3 + \
    4+5+6
print(total)
```

```python
## multiple statements on a single line - You can write multiple statements on a single line by separating them with semicolons (;). However, it's generally recommended to use separate lines for better readability.
x = 5; y = 10; z = x+y
print(z)
```

```python
## Understanding semantics in Python - Python has a simple and readable syntax that emphasizes code readability. It uses indentation to define blocks of code, which helps in understanding the structure of the program. Python also supports dynamic typing, meaning you don't need to declare variable types explicitly.
# variable assignment
age = 32 ## age is an integer
name = "John" ## name is a string
type(age)  # Output: <class 'int'>
type(name)  # Output: <class 'str'>
```

```python
## Type Inference - Python automatically infers the type of a variable based on the value assigned to it. You don't need to explicitly declare the type of a variable.
var=10
print(type(var))  
var="Hello"
print(type(var))
```
