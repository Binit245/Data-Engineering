```python
## if statement

age=20

if age>=18:
    print("You are eligible to vote")
```

```python
## else statement
## The else statement is used to execute a block of code when the condition in the if statement is False.

age=16

if age>=18:
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")
```

```python
## elif statement
## The elif statement is used to check multiple conditions. It stands for "else if" and allows you to specify additional conditions to check if the previous conditions are False.

age=17

if age<13:
    print("You are a child")
elif age<18:
    print("You are a teenager")
else:
    print("You are an adult")
```

```python
## Nested Conditional Statements
## You can place one or more if, elif, or else statements inside another if, elif, or else statement to create nested conditional statements.
 
## number even, odd, negative, positive

num=int(input("Enter the number:"))

if num>0:
    print("The number is positive")
    if num%2==0:
        print("The number is even")
    else:
        print("The number is odd")
else:
    print("The number is zero or negative")

```

```python
## Practical Example

## Determine if the year is leap year using nested conditional statements.

year=int(input("Enter the year: "))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"is a leap year")
        else:
            print(year,"is not a leap year")
    else:
        print(year,"is a leap year")
else:
    print(year,"is not a leap year")
```

```python
## Assignment
## Simple Calculator using nested conditional statements.

num1= int(input("Enter the first number: "))
num2= int(input("Enter the second number: "))
operation= input("Enter operations (+, -, *, /): ")

# Perform the requested operation based on the user's input
if operation == '+':
    result = num1+num2
elif operation == '-':
    result = num1-num2
elif operation == '*':
    result = num1*num2
elif operation == '/':
    if num2 !=0:
        result = num1/num2
    else:
        result = "Error! Division by zero."
else:
    result = "Invalid Operation."

print("Result:", result)
```
